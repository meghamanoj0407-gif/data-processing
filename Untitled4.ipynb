## Data Preprocessing Workflow for Used Car Resale Dataset

### 1. Load the Dataset

```python
import pandas as pd

# Load the dataset
# IMPORTANT: Update 'used_car_resale_dataset.csv' with the actual path to your dataset if it's different.
try:
    df = pd.read_csv('used_car_resale_dataset.csv')
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print("Error: 'used_car_resale_dataset.csv' not found. Please ensure the file is in the correct directory or update the file path.")
    df = None # Set df to None to prevent further errors if file is not found
```
